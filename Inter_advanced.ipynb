```python
#6 task Intermediate level:
print("========== GROCERY SHOPPING BILL ==========")

total_products = int(input("Enter total number of products: "))

bill_items = []
grand_total = 0

for i in range(1, total_products + 1):
    print(f"\n--- Product {i} ---")
    name = input("Product name: ")
    price = float(input("Price Rs: "))
    quantity = int(input("Quantity: "))

    subtotal = price * quantity

    bill_items.append({
        "name": name,
        "price": price,
        "quantity": quantity,
        "subtotal": subtotal
    })

    grand_total += subtotal

# Discount: 10% if bill >= 5000
discount = 0
if grand_total >= 5000:
    discount = grand_total * 0.10

final_amount = grand_total - discount

# Final Receipt
print("\n========== FINAL RECEIPT ==========")
print(f"{'Item':<15} {'Qty':<5} {'Price':<10} {'Subtotal'}")
print("-----------------------------------")
for item in bill_items:
    print(f"{item['name']:<15} {item['quantity']:<5} Rs.{item['price']:<8} Rs.{item['subtotal']:.2f}")

print("-----------------------------------")
print(f"Subtotal:             Rs.{grand_total:.2f}")
print(f"Discount 10%:         Rs.{discount:.2f}")
print(f"Final Amount to Pay:  Rs.{final_amount:.2f}")
print("===================================")
```

```python

#7 task

name = input("Enter Name: ")
hours = float(input("Enter Hours: "))
rate = float(input("Enter Rate: "))

normal = 40

if hours > normal:
    normal_pay = normal * rate
    ot_hours = hours - normal
    ot_pay = ot_hours * rate * 1.5
else:
    normal_pay = hours * rate
    ot_pay = 0

total = normal_pay + ot_pay

print("\n--- Salary Slip ---")
print("Name:", name)
print("Hours:", hours)
print("Normal Pay:", normal_pay)
print("Overtime Pay:", ot_pay)
print("Total Salary:", total)
```

```python
#task 15 Clinic Appointment Manager
patients = []

def add_patient():
    patient_id = input("Enter Patient ID: ")

    # Check if Patient ID already exists
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient already exists!")
            return

    patient_name = input("Enter Patient Name: ")
    patient_age = int(input("Enter Age: "))
    problem = input("Enter Disease/Problem: ")
    time = input("Enter Appointment Time: ")
    emergency = input("Is this an emergency patient? (yes/no): ")

    # Store patient with all details
    patient = {
        "id": patient_id,
        "name": patient_name,
        "age": patient_age,
        "problem": problem,
        "time": time,
        "emergency": emergency
    }

    patients.append(patient)
    print("Appointment added successfully")

def search_patient():
    patient_id = input("Enter Patient ID to search: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("\nPatient Found!")
            print("Patient ID:", patient['id'])      # ID
            print("Name:", patient['name'])          # Name
            print("Age:", patient['age'])            # Age
            print("Disease:", patient['problem'])    # Disease
            print("Appointment Time:", patient['time']) # Time
            if patient['emergency'].lower() == 'yes':
                print("Status: Emergency")
            else:
                print("Status: Normal")
            return
    print("Patient not found!")

def display_patients():
    if len(patients) == 0:
        print("No patients found!")
        return

    print("\n========= All Appointments =========")

    # Show Emergency patients first
    for patient in patients:
        if patient["emergency"].lower() == "yes":
            print("Patient ID:", patient['id'])
            print("Name:", patient['name'])
            print("Age:", patient['age'])
            print("Disease:", patient['problem'])
            print("Appointment Time:", patient['time'])
            print("Status: Emergency")
            print("------------------------")

    # Then show Normal patients
    for patient in patients:
        if patient["emergency"].lower() == "no":
            print("Patient ID:", patient['id'])
            print("Name:", patient['name'])
            print("Age:", patient['age'])
            print("Disease:", patient['problem'])
            print("Appointment Time:", patient['time'])
            print("Status: Normal")
            print("------------------------")

def cancel_appointment():
    patient_id = input("Enter Patient ID to cancel appointment: ")

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Appointment cancelled successfully!")
            return
    print("Patient not found!")

def count_patients():
    emergency_count = 0
    normal_count = 0

    for patient in patients:
        if patient["emergency"].lower() == "yes":
            emergency_count += 1
        else:
            normal_count += 1

    print("\n========= Clinic Summary =========")
    print("Total Normal patients:", normal_count)
    print("Total Emergency patients:", emergency_count)
    print("Total patients:", len(patients))

# Main Menu
while True:
    print("\n========= Clinic Appointment Manager =========")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_patient()
    elif choice == "2":
        search_patient()
    elif choice == "3":
        display_patients()
    elif choice == "4":
        cancel_appointment()
    elif choice == "5":
        count_patients()
    elif choice == "6":
        print("Program Ended")
        break
    else:
        print("Invalid choice!")
```

```python
# Task 14: Restaurant Ordering System

menu= {
    "Burger":1500,
    "Pizza":2000,
    "Pasta":1200,
    "Drink":550
}

orders = []

print("========= Restaurant Menu =========")
for key,value in menu.items():
  print(f'{key} : {value}')

while True:
  item_name = input("Enter your item name: ").title()
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
    print("This item is not available")

  choice = input("You want to add more items: yes/no: ")
  if choice == "no":
    break

total = 0
for order in orders:
  total = total + order["sub_total"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1
else:
  discount = 0

grand_total = total + tax - discount

def receipt():
  print("\n============ Receipt ============")

  for order in orders:
    print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')

  print("Total: ", total)
  print("Tax: ", tax)
  print("Discount: ", discount)
  print("Grand Total: ", grand_total)

receipt()
```
