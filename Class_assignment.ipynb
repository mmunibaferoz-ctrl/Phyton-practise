```python
#5 secure pin checker
secret_pin = 456789
for i in range (3):
  user_input = int(input("enter your pin : "))
  if user_input == secret_pin:
   print("access granted")

  else:
    print("access denied")
    print(f"no of attempts {i}")
    user_input = int(input("enter your pin"))
    print("Account blocked")
```

```python

#Task 1 cinema ticket price

name = input("enter your name")
age = int(input("enter your age:"))
if age < 5 or age > 100:
    ticket_price = "Free"
elif age >= 5 and age <= 12:
    ticket_price = 300
elif age >= 13 and age <= 59:
    ticket_price = 700
elif age >= 60 and age <= 100:
    ticket_price = 400

print("customer:", name)
print("Ticket:", ticket_price)
```

```python
#task 4 weekly step checker
total_steps = 0
user_input = int(input("enter your daily steps: "))

for day in range(6):
    user_input = int(input("enter your daily steps: "))
    total_steps = total_steps + user_input
    daily_avg = total_steps / 7
print("total steps:", total_steps)
print("daily_average:", daily_avg)
if daily_avg >= 5000:
    print("Active Week")
else:
    print("Needs More Activity")
```

```python
#task 3 mobile recharge calculator

mobile = input("enter mobile number:")
recharge = float(input("enter recharge amount:"))
tax = recharge * 3 / 100
Final_balanced = recharge - tax
print("mobile number:", mobile)
print("recharge amount:", recharge)
print("tax deducted:", Final_balanced)
```

```python

#task 2: temperature converter
Celsius = float(input("enter a temperature in celsius:"))

#formula
Fahrenheit =(Celsius * 9/5) + 32

print(Celsius, "°C =", Fahrenheit, "°F")

if Celsius <= 10 :
   print("weather: very cold")
elif Celsius <= 19 :
   print("weather:cold")
elif Celsius <= 29 :
   print("weather:normal")

else:
    print("weather:hot")
```
