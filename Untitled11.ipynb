```python
import pandas as pd
```

```python


# Task 1: DataFrame banana
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)
print(df)

# Find
print("\n--- Results ---")

# 1. Average marks
avg_marks = df["Marks"].mean()
print(f"Average marks: {avg_marks}")

# 2. Maximum marks
max_marks = df["Marks"].max()
print(f"Maximum marks: {max_marks}")

# 3. Minimum marks
min_marks = df["Marks"].min()
print(f"Minimum marks: {min_marks}")

# 4. Total students
total_students = len(df)
print(f"Total students: {total_students}")
```

```python

# Task 2:
# (Pass)
pass_students = df[df["Marks"] >= 50]
print("Students with marks >= 50:")
print(pass_students)

print()

#  (Fail)
fail_students = df[df["Marks"] < 50]
print("Students with marks < 50:")
print(fail_students)
```

```python
# Task 3:

# 1. City
df["City"] = ["Karachi", "Lahore", "Karachi", "Islamabad", "Quetta"]

print("Updated DataFrame:")
print(df)

print

# 2.
karachi_students = df[df["City"] == "Karachi"]
print("Students from Karachi:")
print(karachi_students)
```

```python
import pandas as pd
import numpy as np

# Task 5: DataFrame with missing marks
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [85, np.nan, 75, np.nan, 90]
}

df = pd.DataFrame(data)
print()
print(df)

print()

# 1.
missing_count = df["Marks"].isna().sum()
print(f"Number of missing values: {missing_count}")

# 2.
avg_marks = df["Marks"].mean()
print(f"Average marks: {avg_marks}")

df["Marks"] = df["Marks"].fillna(avg_marks)

# 3.
print("\nCleaned DataFrame:")
print(df)
```
